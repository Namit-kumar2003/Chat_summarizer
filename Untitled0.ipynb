```python
train_df = pd.read_csv("/content/samsum_train.csv")
train_df = train_df.dropna(subset=["dialogue", "summary"])

print(f"Dataset size: {len(train_df)} samples")
print(f"Sample dialogue:\n{train_df['dialogue'].iloc[0][:200]}...")
print(f"\nSample summary:\n{train_df['summary'].iloc[0]}")
```

```python
def clean_dialogue(dialogue):
    """Minimal cleaning - only remove empty lines and excess whitespace"""
    lines = dialogue.split("\n")
    keep = []
    for l in lines:
        stripped = l.strip()
        if len(stripped) > 0:
            keep.append(stripped)
    return "\n".join(keep)

train_df["dialogue"] = train_df["dialogue"].apply(clean_dialogue)
print("\n✓ Data cleaning complete")

# Show sample to verify names are present
print("\n📝 Sample dialogue (check for real names):")
print(train_df['dialogue'].iloc[0][:300])
print(f"\n📝 Sample summary:")
print(train_df['summary'].iloc[0])
```

```python
def format_input(dialogue):

    return (
        "Summarize this conversation using the actual person names mentioned. "
        "Do not use A, B, C labels:\n\n" + dialogue
    )
```

```python
dataset = Dataset.from_pandas(train_df)
dataset = dataset.train_test_split(test_size=0.1, seed=42)

print(f"\nTrain samples: {len(dataset['train'])}")
print(f"Validation samples: {len(dataset['test'])}")
```

```python
model_name = "philschmid/bart-large-cnn-samsum"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
print(f"✓ Model loaded on {device}")
```

```python
def tokenize(batch):
    """Tokenize with -100 masking for padded labels"""
    inputs = tokenizer(
        [format_input(d) for d in batch["dialogue"]],
        padding="max_length",
        truncation=True,
        max_length=512
    )

    labels = tokenizer(
        batch["summary"],
        padding="max_length",
        truncation=True,
        max_length=128
    )

    # CRITICAL: Replace pad tokens with -100 for proper loss calculation
    labels_ids = labels["input_ids"]
    labels_ids = [
        [(l if l != tokenizer.pad_token_id else -100) for l in label]
        for label in labels_ids
    ]

    inputs["labels"] = labels_ids
    # Store original summaries for evaluation
    inputs["summary_text"] = batch["summary"]
    return inputs

print("\n🔄 Tokenizing datasets...")
train_data = dataset["train"].map(
    tokenize,
    batched=True,
    remove_columns=["dialogue"],  # Keep summary_text column
    desc="Tokenizing train"
)

val_data = dataset["test"].map(
    tokenize,
    batched=True,
    remove_columns=["dialogue"],  # Keep summary_text column
    desc="Tokenizing validation"
)

print("✓ Tokenization complete")
```

```python
def collate_fn(batch):
    return {
        "input_ids": torch.tensor([x["input_ids"] for x in batch]).to(device),
        "attention_mask": torch.tensor([x["attention_mask"] for x in batch]).to(device),
        "labels": torch.tensor([x["labels"] for x in batch]).to(device),
        "summary_text": [x["summary_text"] for x in batch]  # Keep text summaries
    }

# Use larger batch size (reduce if OOM)
BATCH_SIZE = 4

train_loader = DataLoader(
    train_data,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_fn
)
val_loader = DataLoader(
    val_data,
    batch_size=BATCH_SIZE,
    collate_fn=collate_fn
)

print(f"\n✓ DataLoaders ready (batch size: {BATCH_SIZE})")
```

```python
def evaluate_model(model, val_loader, tokenizer, device, max_samples=None):
    """Evaluate model and return ROUGE scores
    Args:
        max_samples: If set, only evaluate on first N samples for speed
    """
    rouge = evaluate.load("rouge")
    model.eval()
    preds, refs = [], []

    print("\n📈 Evaluating model...")
    sample_count = 0

    with torch.no_grad():
        for batch in tqdm(val_loader, desc="Generating summaries"):
            input_ids = batch["input_ids"]
            summary_texts = batch["summary_text"]

            outputs = model.generate(
                input_ids,
                num_beams=4,
                min_length=20,
                max_length=128,
                length_penalty=1.0,
                no_repeat_ngram_size=3,
                early_stopping=True,
                num_return_sequences=1  # Ensure single output
            )

            # Decode predictions
            batch_preds = tokenizer.batch_decode(outputs, skip_special_tokens=True)

            preds.extend(batch_preds)
            refs.extend(summary_texts)

            sample_count += len(batch_preds)
            if max_samples and sample_count >= max_samples:
                print(f"   Evaluated {sample_count} samples (limited for speed)")
                break

    scores = rouge.compute(predictions=preds, references=refs)
    return scores, preds, refs
```

```python
print("\n" + "="*60)
print("BASELINE EVALUATION (Zero-shot - No Fine-tuning)")
print("="*60)

# Fast baseline: evaluate on subset first
baseline_scores, baseline_preds, baseline_refs = evaluate_model(
    model, val_loader, tokenizer, device, max_samples=200
)

print(f"\n📊 Baseline ROUGE Scores (on 200 samples):")
print(f"   ROUGE-1: {baseline_scores['rouge1']:.4f}")
print(f"   ROUGE-2: {baseline_scores['rouge2']:.4f}")
print(f"   ROUGE-L: {baseline_scores['rougeL']:.4f}")

print(f"\n📝 Sample predictions:")
for i in range(min(3, len(baseline_preds))):
    print(f"\nExample {i+1}:")
    print(f"Reference: {baseline_refs[i]}")
    print(f"Predicted: {baseline_preds[i]}")
```

```python
def evaluate_model(model, val_loader, tokenizer, device, max_samples=None):
    """Evaluate model and return ROUGE scores
    Args:
        max_samples: If set, only evaluate on first N samples for speed
    """
    rouge = evaluate.load("rouge")
    model.eval()
    preds, refs = [], []

    print("\n📈 Evaluating model...")
    sample_count = 0

    with torch.no_grad():
        for batch in tqdm(val_loader, desc="Generating summaries"):
            input_ids = batch["input_ids"]
            summary_texts = batch["summary_text"]

            outputs = model.generate(
                input_ids,
                num_beams=4,
                min_length=20,
                max_length=128,
                length_penalty=1.0,
                no_repeat_ngram_size=3,
                early_stopping=True,
                num_return_sequences=1  # Ensure single output
            )

            # Decode predictions
            batch_preds = tokenizer.batch_decode(outputs, skip_special_tokens=True)

            preds.extend(batch_preds)
            refs.extend(summary_texts)

            sample_count += len(batch_preds)
            if max_samples and sample_count >= max_samples:
                print(f"   Evaluated {sample_count} samples (limited for speed)")
                break

    scores = rouge.compute(predictions=preds, references=refs)
    return scores, preds, refs
```

```python
print("\n" + "="*60)
print("BASELINE EVALUATION (Zero-shot - No Fine-tuning)")
print("="*60)

# Fast baseline: evaluate on subset first
baseline_scores, baseline_preds, baseline_refs = evaluate_model(
    model, val_loader, tokenizer, device, max_samples=200
)

print(f"\n📊 Baseline ROUGE Scores (on 200 samples):")
print(f"   ROUGE-1: {baseline_scores['rouge1']:.4f}")
print(f"   ROUGE-2: {baseline_scores['rouge2']:.4f}")
print(f"   ROUGE-L: {baseline_scores['rougeL']:.4f}")

print(f"\n📝 Sample predictions:")
for i in range(min(3, len(baseline_preds))):
    print(f"\nExample {i+1}:")
    print(f"Reference: {baseline_refs[i]}")
    print(f"Predicted: {baseline_preds[i]}")
```

```python
LEARNING_RATE = 2e-5  # Slightly higher to adapt faster
NUM_EPOCHS = 3  # Increased from 3 to 5
WARMUP_STEPS = 100

optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE)

# Learning rate scheduler with warmup
from torch.optim.lr_scheduler import OneCycleLR

total_steps = len(train_loader) * NUM_EPOCHS
scheduler = OneCycleLR(
    optimizer,
    max_lr=LEARNING_RATE,
    total_steps=total_steps,
    pct_start=0.1,
    anneal_strategy='cos'
)

print("\n" + "="*60)
print("FINE-TUNING CONFIGURATION")
print("="*60)
print(f"Learning Rate: {LEARNING_RATE}")
print(f"Epochs: {NUM_EPOCHS}")
print(f"Batch Size: {BATCH_SIZE}")
print(f"Total Training Steps: {total_steps}")
```

```python
print("\n🚀 Starting fine-tuning...")
print("="*60)

model.train()
best_rouge = baseline_scores['rougeL']

for epoch in range(NUM_EPOCHS):
    print(f"\n📚 Epoch {epoch+1}/{NUM_EPOCHS}")
    total_loss = 0

    progress_bar = tqdm(train_loader, desc=f"Training Epoch {epoch+1}")

    for batch in progress_bar:
        # Remove summary_text from batch for training
        train_batch = {
            "input_ids": batch["input_ids"],
            "attention_mask": batch["attention_mask"],
            "labels": batch["labels"]
        }

        outputs = model(**train_batch)
        loss = outputs.loss

        loss.backward()

        # Gradient clipping for stability
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

        optimizer.step()
        scheduler.step()
        optimizer.zero_grad()

        total_loss += loss.item()
        progress_bar.set_postfix({'loss': f'{loss.item():.4f}'})

    avg_loss = total_loss / len(train_loader)
    print(f"   Average Loss: {avg_loss:.4f}")

    # Evaluate after each epoch (on subset for speed during training)
    epoch_scores, epoch_preds, epoch_refs = evaluate_model(
        model, val_loader, tokenizer, device, max_samples=200
    )

    print(f"\n   📊 Epoch {epoch+1} ROUGE Scores (on 200 samples):")
    print(f"      ROUGE-1: {epoch_scores['rouge1']:.4f}")
    print(f"      ROUGE-2: {epoch_scores['rouge2']:.4f}")
    print(f"      ROUGE-L: {epoch_scores['rougeL']:.4f}")

    # Save best model
    if epoch_scores['rougeL'] > best_rouge:
        best_rouge = epoch_scores['rougeL']
        print(f"   ⭐ New best ROUGE-L: {best_rouge:.4f}")
        # Uncomment to save model
        # model.save_pretrained("./best_model")
        # tokenizer.save_pretrained("./best_model")

    model.train()
```

```python
print("\n" + "="*60)
print("FINAL EVALUATION (Full validation set)")
print("="*60)

# Full evaluation at the end
final_scores, final_preds, final_refs = evaluate_model(
    model, val_loader, tokenizer, device, max_samples=None  # Evaluate all
)

print(f"\n🎯 FINAL ROUGE SCORES:")
print(f"   ROUGE-1: {final_scores['rouge1']:.4f}")
print(f"   ROUGE-2: {final_scores['rouge2']:.4f}")
print(f"   ROUGE-L: {final_scores['rougeL']:.4f}")

print(f"\n📈 IMPROVEMENT:")
print(f"   ROUGE-L: {baseline_scores['rougeL']:.4f} → {final_scores['rougeL']:.4f}")
print(f"   Gain: +{(final_scores['rougeL'] - baseline_scores['rougeL']):.4f}")

if final_scores['rougeL'] >= 0.60:
    print(f"\n🎉 SUCCESS! Target ROUGE-L of 0.60+ achieved!")
else:
    print(f"\n⚠️ Target not reached. Current: {final_scores['rougeL']:.4f}")
    print("   Try: Increase epochs to 5, adjust batch size, or check data quality")
```

```python
# ============================================================================
# FINAL KAGGLE SUBMISSION FILE GENERATOR
# Run this AFTER your training pipeline completes
# ============================================================================

import pandas as pd
import torch
from tqdm import tqdm
import re

print("\n" + "="*70)
print("🎯 GENERATING FINAL KAGGLE SUBMISSION FILE")
print("="*70)

# ============================================================================
# STEP 1: VERIFY MODEL IS LOADED
# ============================================================================

print("\n🔍 Checking if model is available...")
try:
    # Test if variables exist
    test_model = model
    test_tokenizer = tokenizer
    test_device = device
    print("✅ Model, tokenizer, and device are available from training pipeline")
    print(f"   Device: {device}")
except NameError as e:
    print(f"❌ ERROR: {e}")
    print("\n⚠️  CRITICAL: You must run the training pipeline first!")
    print("   The model needs to be in memory from your training.")
    raise Exception("Model not found. Please run training pipeline first.")

# ============================================================================
# STEP 2: LOAD TEST DATA
# ============================================================================

print("\n📂 Loading test data...")
test_df = pd.read_csv("/content/samsum_test.csv")
print(f"✅ Loaded {len(test_df)} test samples")

# Verify required columns
if 'dialogue' not in test_df.columns:
    raise Exception("ERROR: 'dialogue' column not found in test file!")

if 'id' not in test_df.columns:
    print("⚠️  No 'id' column found, creating sequential IDs...")
    test_df['id'] = range(len(test_df))

print(f"   Columns: {list(test_df.columns)}")

# ============================================================================
# STEP 3: PREPROCESSING (SAME AS TRAINING)
# ============================================================================

print("\n🔄 Preprocessing test data...")

def clean_dialogue(dialogue):
    """Minimal cleaning - same as training"""
    if pd.isna(dialogue):
        return ""
    lines = str(dialogue).split("\n")
    keep = []
    for l in lines:
        stripped = l.strip()
        if len(stripped) > 0:
            keep.append(stripped)
    return "\n".join(keep)

def format_input(dialogue):
    """Same prompt format as training"""
    return (
        "Summarize this conversation using the actual person names mentioned. "
        "Do not use A, B, C labels:\n\n" + dialogue
    )

# Apply preprocessing
test_df["dialogue"] = test_df["dialogue"].apply(clean_dialogue)
print(f"✅ Preprocessing complete")

# Show sample
print(f"\n📝 Sample test dialogue:")
print(test_df['dialogue'].iloc[0][:200] + "...")

# ============================================================================
# STEP 4: BATCH SUMMARY GENERATION
# ============================================================================

def generate_summaries_batch(model, dialogues, tokenizer, device, batch_size=8):
    """
    Generate summaries in batches for efficiency

    Args:
        model: Fine-tuned model
        dialogues: List of dialogue strings
        tokenizer: Tokenizer
        device: torch device
        batch_size: Number of dialogues to process at once

    Returns:
        List of generated summaries
    """
    model.eval()
    all_summaries = []

    with torch.no_grad():
        for i in tqdm(range(0, len(dialogues), batch_size), desc="Generating summaries"):
            # Get batch
            batch_dialogues = dialogues[i:i + batch_size]

            # Format inputs
            formatted = [format_input(d) for d in batch_dialogues]

            # Tokenize
            inputs = tokenizer(
                formatted,
                padding=True,
                truncation=True,
                max_length=512,
                return_tensors="pt"
            ).to(device)

            # Generate
            outputs = model.generate(
                inputs["input_ids"],
                attention_mask=inputs["attention_mask"],
                num_beams=4,
                min_length=20,
                max_length=128,
                length_penalty=1.0,
                no_repeat_ngram_size=3,
                early_stopping=True,
                num_return_sequences=1
            )

            # Decode
            batch_summaries = tokenizer.batch_decode(outputs, skip_special_tokens=True)
            all_summaries.extend(batch_summaries)

    return all_summaries

print(f"\n🤖 Generating summaries for {len(test_df)} dialogues...")
print(f"   This may take a few minutes...")

summaries = generate_summaries_batch(
    model=model,
    dialogues=test_df["dialogue"].tolist(),
    tokenizer=tokenizer,
    device=device,
    batch_size=8  # Adjust if you get OOM errors
)

print(f"✅ Generated {len(summaries)} summaries")

# ============================================================================
# STEP 5: POST-PROCESSING
# ============================================================================

print("\n🔧 Post-processing summaries...")

def post_process_summary(summary):
    """
    Clean and format generated summaries
    - Capitalize first letter
    - Add period if missing
    - Remove extra whitespace
    - Remove any A:, B:, C: artifacts
    """
    if not summary or pd.isna(summary):
        return ""

    summary = str(summary).strip()

    # Remove any speaker labels that leaked through
    summary = re.sub(r'\b[A-Z]:\s*', '', summary)
    summary = re.sub(r'\s+[A-Z]:\s+', ' ', summary)

    # Remove extra whitespace
    summary = re.sub(r'\s+', ' ', summary).strip()

    # Capitalize first letter
    if summary and len(summary) > 0:
        summary = summary[0].upper() + summary[1:]

    # Ensure ends with punctuation
    if summary and summary[-1] not in ['.', '!', '?']:
        summary += '.'

    return summary

test_df["summary"] = [post_process_summary(s) for s in summaries]
print("✅ Post-processing complete")

# ============================================================================
# STEP 6: VALIDATION CHECKS
# ============================================================================

print("\n🔍 Validating submission...")

# Check for missing summaries
missing = test_df["summary"].isna().sum()
empty = (test_df["summary"].str.strip() == "").sum()

if missing > 0:
    print(f"   ⚠️  WARNING: {missing} missing summaries!")
    # Fill with placeholder
    test_df["summary"].fillna("No summary available.", inplace=True)

if empty > 0:
    print(f"   ⚠️  WARNING: {empty} empty summaries!")
    # Fill with placeholder
    test_df.loc[test_df["summary"].str.strip() == "", "summary"] = "No summary available."

# Check for duplicates
duplicates = test_df["id"].duplicated().sum()
if duplicates > 0:
    print(f"   ⚠️  WARNING: {duplicates} duplicate IDs detected!")

# Summary statistics
avg_length = test_df["summary"].str.len().mean()
min_length = test_df["summary"].str.len().min()
max_length = test_df["summary"].str.len().max()

print(f"\n📊 Summary Statistics:")
print(f"   Total samples: {len(test_df)}")
print(f"   Average length: {avg_length:.1f} characters")
print(f"   Min length: {min_length} characters")
print(f"   Max length: {max_length} characters")
print(f"   Missing/Empty: {missing + empty}")

# ============================================================================
# STEP 7: CREATE SUBMISSION FILE
# ============================================================================

print("\n💾 Creating submission.csv...")

# Select only required columns in correct order
submission_df = test_df[["id", "summary"]].copy()

# Ensure id is integer type
submission_df["id"] = submission_df["id"].astype(int)

# Save to CSV
submission_df.to_csv("/content/submission.csv", index=False)

print(f"✅ submission.csv created successfully!")
print(f"   Location: /content/submission.csv")
print(f"   Rows: {len(submission_df)}")
print(f"   Columns: {list(submission_df.columns)}")

# ============================================================================
# STEP 8: DISPLAY SAMPLE PREDICTIONS
# ============================================================================

print("\n" + "="*70)
print("📝 SAMPLE PREDICTIONS (First 5)")
print("="*70)

for i in range(min(5, len(test_df))):
    print(f"\n{'='*70}")
    print(f"Sample {i+1} (ID: {test_df['id'].iloc[i]})")
    print(f"{'='*70}")
    print(f"Dialogue:\n{test_df['dialogue'].iloc[i][:300]}...")
    print(f"\n→ Generated Summary:\n{test_df['summary'].iloc[i]}")

# ============================================================================
# STEP 9: SAVE DETAILED VERSION (OPTIONAL)
# ============================================================================

print("\n💾 Saving detailed version for analysis...")
detailed_df = test_df[["id", "dialogue", "summary"]].copy()
detailed_df.to_csv("/content/submission_detailed.csv", index=False)
print("✅ submission_detailed.csv saved")

# ============================================================================
# FINAL MESSAGE
# ============================================================================

print("\n" + "="*70)
print("🎉 SUBMISSION FILE READY FOR KAGGLE!")
print("="*70)
print(f"\n✅ File: /content/submission.csv")
print(f"✅ Format: id, summary")
print(f"✅ Samples: {len(submission_df)}")
print(f"\n📥 TO DOWNLOAD:")
print("   1. Click folder icon 📁 on the left")
print("   2. Find 'submission.csv'")
print("   3. Right-click → Download")
print("   4. Upload to Kaggle")
print(f"\n🎯 Expected Kaggle ROUGE Score: ~0.41-0.42")
print("   (Based on validation score: 0.4183)")
print("\n" + "="*70)

# Quick verification
print("\n🔍 Quick File Verification:")
verify_df = pd.read_csv("/content/submission.csv")
print(f"   ✅ File readable: {len(verify_df)} rows")
print(f"   ✅ Columns present: {list(verify_df.columns)}")
print(f"   ✅ No null summaries: {verify_df['summary'].isna().sum() == 0}")
print("\n✨ All checks passed! Ready for submission! ✨")
```
